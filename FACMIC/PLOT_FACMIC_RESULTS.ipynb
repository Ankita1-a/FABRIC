## **FACMIC- Reproduced result with same hyperparameters as our modified approach**

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
acc = pd.read_csv('/content/drive/MyDrive/FACMIC/results/Global_Acc7_new.csv', header=None).squeeze()
bacc = pd.read_csv('/content/drive/MyDrive/FACMIC/results/Global_bacc7_new.csv', header=None).squeeze()

epochs = range(1, len(acc) + 1)

# --- Print stats ---
for name, data in [("Accuracy", acc), ("Balanced Accuracy", bacc)]:
    best_epoch = data.idxmax() + 1  # 1-indexed
    best_val = data.max()
    first_val = data.iloc[0]
    final_val = data.iloc[-1]
    print(f"=== {name} ===")
    print(f"  First  (Epoch  1): {first_val:.6f}")
    print(f"  Best   (Epoch {best_epoch:2d}): {best_val:.6f}")
    print(f"  Final  (Epoch {len(data):2d}): {final_val:.6f}")
    print()

# --- Plot ---
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
fig.suptitle("Accuracy and Balanced Accuracy for BrainTumor using FACMIC", fontsize=14, fontweight='bold')

configs = [
    ("Accuracy",          acc,  "#2563EB"),
    ("Balanced Accuracy", bacc, "#16A34A"),
]

for ax, (name, data, color) in zip(axes, configs):
    best_epoch = data.idxmax() + 1
    best_val = data.max()

    ax.plot(epochs, data, color=color, linewidth=2, label=name)
    ax.axvline(best_epoch, color='red', linestyle='--', linewidth=1, alpha=0.7, label=f'Best epoch ({best_epoch})')
    ax.scatter([best_epoch], [best_val], color='red', zorder=5)
    ax.annotate(f"  Best: {best_val:.4f}\n  Epoch {best_epoch}",
                xy=(best_epoch, best_val),
                fontsize=8.5, color='red',
                va='bottom')

    ax.set_title(name, fontsize=12)
    ax.set_xlabel("Epoch")
    ax.set_ylabel(name)
    ax.set_xlim(1, len(data))
    ax.legend(fontsize=9)
    ax.grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.savefig('/content/drive/MyDrive/FACMIC/results/metrics_plot_new.png', dpi=150, bbox_inches='tight')
plt.show()
print("Plot saved to metrics_plot_new.png")
```
